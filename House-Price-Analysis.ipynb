```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Area_sqft": [850, 1200, 1500, 1800, 2200, 2500, 3000, 1100, 1600, 2000],
    "Bedrooms": [2, 2, 3, 3, 4, 4, 5, 2, 3, 4],
    "Bathrooms": [1, 2, 2, 3, 3, 3, 4, 2, 2, 3],
    "Age_years": [10, 8, 5, 3, 2, 1, 4, 7, 6, 3],
    "Price_Lakhs": [35, 50, 65, 80, 100, 120, 150, 45, 70, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
for column in df.columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Area_sqft"], df["Price_Lakhs"])

plt.xlabel("Area (sqft)")
plt.ylabel("Price (Lakhs)")
plt.title("House Area vs Price")

plt.show()
```

```python
bedroom_price = df.groupby("Bedrooms")["Price_Lakhs"].mean()

print("Average House Price by Number of Bedrooms:")
print(bedroom_price)

plt.figure(figsize=(8, 5))
bedroom_price.plot(kind="bar")

plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Price (Lakhs)")
plt.title("Average House Price by Bedrooms")

plt.show()
```

```python
bathroom_price = df.groupby("Bathrooms")["Price_Lakhs"].mean()

print("Average House Price by Bathrooms:")
print(bathroom_price)

plt.figure(figsize=(8, 5))
bathroom_price.plot(kind="bar")

plt.xlabel("Number of Bathrooms")
plt.ylabel("Average Price (Lakhs)")
plt.title("Average House Price by Bathrooms")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)

print("Correlation with House Price:")
print(correlation["Price_Lakhs"].sort_values(ascending=False))

plt.figure(figsize=(8, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Price_Lakhs"], bins=5, edgecolor="black")

plt.xlabel("House Price (Lakhs)")
plt.ylabel("Number of Houses")
plt.title("Distribution of House Prices")

plt.show()
```

```python
most_expensive = df.loc[df["Price_Lakhs"].idxmax()]
cheapest = df.loc[df["Price_Lakhs"].idxmin()]

print("Most Expensive House:")
print(most_expensive)

print("\nCheapest House:")
print(cheapest)
```

```python
average_price = df["Price_Lakhs"].mean()
average_area = df["Area_sqft"].mean()
average_bedrooms = df["Bedrooms"].mean()

print("House Price Analysis")
print("--------------------")
print("Average House Price:", round(average_price, 2), "Lakhs")
print("Average Area:", round(average_area, 2), "sqft")
print("Average Bedrooms:", round(average_bedrooms, 2))

print("\nKey Insights:")
print("1. Larger houses generally have higher prices.")
print("2. Houses with more bedrooms tend to have higher prices.")
print("3. The correlation heatmap shows relationships between features.")
print("4. House prices vary across the dataset.")
```
